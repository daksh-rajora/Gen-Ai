#### plt.show()---> helps in representing the chart

## Customization For Line
#### Parameters:
#### 1. color  ---> What is going to be the color of the line --> for eg: red,greeen, blue, or HTML hash code

#### 2. marker ---> how point of intersection appears --> o, s, +, *, d
#### 3. linestyle ---> it will be defining the style  of the line ---> '--','-.'
#### 4. linewidth ---> it shows the width of the line ---> 1,1.5,2
#### 5. markersize---> it shows the markersize or it can change the size of the marker, default-->6, --> 10,9,12
#### 6. label --> label that what this specific line is representing but is only helpful while plotting multiple lines in a single visualization

```python
data=data.sort_index()
```

```python
x=data.index
```

```python
y=data.values
```

```python
plt.figure(figsize=(10,6))
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.plot(x,y,linewidth=1.5,linestyle='--',color='red',marker='o')
plt.grid()
plt.show()
```

## Types of Plots
#### 1. Line Plot ---> Time Series Analysis ---> How values are varying as per Time
#### 2. Scatter Plot --> Scatter Plot is being used for Bivariate Analysis ---> Two Numerical Columns --> We check that based on the changes in one Numerical Column Value ---> How it affects the other Numerical Column Value
#### for example --->  x ---> Age  ---> 10,15,20,25,30...
####  y --> Cholestrol Level ---> 50,60,75,80,....

#### Scatter Plot ---> It will figure out that based on the increment in Age Value --> If cholestrol level is Increasing or Decreasing or Is Steady
#### Scatter Plot ---> it figures out the correlation between the two numerical values

```python
x=np.linspace(-10,10,50)
y=10+x+3 + np.random.randint(0,300,50)
plt.scatter(x,y)
```
