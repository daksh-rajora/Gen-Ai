## Relational Plot

#### Relational Plot is being used to figure out the relationship between two or more Numerical Variables and It helps us out to figure out that on bringing the changes in one numerical column, How it affects the other Numerical Column

#### Relational Plot ----> Bivariate Analysis
#### ---> Numerical - Numerical
#### ---> One Numerical Column ---> Change ---> How it Affects ---> Other Numerical Column
#### Line Plot ---> Line Plot is being used for Time Series Analysis (wrt Time how one quantity is going to vary) ---> X axis ---> Time Y axis --> Value that will vary
#### For ex --> wrt Time How my weight is varying or how sales value is varying with time or how stocl price for a company is varying with Time
#### Scatterplot --> Scatterplot show how 2 numerical variables vary or are correlated that means if one numerical column will vary then how it will affect the other numerical column
#### Correlation btw 2 Numerical Column
#### Study Hours <---> Exam score
#### Axis Level Functions 
#### Line Plot --> sns.lineplot()
#### lineplot() --->
#### data--> dataframe value from where we will be selecting the columns for analysis
#### x--> 'Column_name' that you want to keep on the X axis and this column name will directly behave as the y label for us
#### hue --->it can group or seprate the information in the plot with respect to a new column
#### style ---> style parameter can also add additional insights in the graph by changing the plot stule (lineplot---> line stype will change)
#### size ---> size of the plot can vary with respect to the column
#### Scatter Plot --> sns.scatterplot()
#### Figure Level Function
#### relplot() ---> it is capable of plotting both scatterplot and lineplot just based on the kind parameter
#### kind='line'---> lineplot
#### kind='scatter'---> scatterplot

```python
# tips

tips=sns.load_dataset('tips')
```

```python
tips.head()
```

```python
## axis level function
sns.lineplot(data=tips,x='day',y='total_bill',hue='smoker',style='sex')
```

```python
## figure level function

sns.relplot(data=tips,x='day',y='total_bill',kind='line',hue='sex',style='smoker')
```
