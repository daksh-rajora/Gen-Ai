## Facet Plot
#### facet plot can help us divide the plot as per the multiple categories of a categorical column such as let us say, we dont want to use hue and we want to divide the plot or graph as per multiple categories of a categorical column as row wise or column wise then we can go with the facet plot
#### facet plot can help us separate the info Rowwise or Columnwise
#### We can use Row or column parameter to mention that if we want to separate the info rowwise or column wise
#### Facet plot functionality is only possible in the Figure level plot
#### As figure level plot provides us the control over the whole figure and axis level plot only provides us the control over the axis itself

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
sns.relplot(data=tips,x='total_bill',row='sex',y='tip',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',col='sex',y='tip',kind='scatter')
```

## Matrics Plot
#### Matrix Plot shows Relationship between Two Dimensions like a Table using colors to show the strength of pattern of data.
#### Matrix Plot can be considered similar to those of the Excel sheets where in Excel Sheet cells instead of numbers, we can use the color intensity to understand values.
#### Heatmap
#### In Every classification Machine Learning Problem Statement, we use the Heatmap.
#### Heatmap uses color shades to show the size of Numers in a Matrix 2D table.
#### Darker or lighter colors indicate higher or lower values.
#### Super useful when we want to see correlation,  Missing Data, Comparison, density
#### for ex ---> Sales over the Months in Different Countries
#### For comming up an Heatmap or an Matrix which we can convert into a Heatmap, we create a Pivot Table and then that pivot table only we can use to create an heatmap
#### pivot()--->index, columns, values ---> gap.pivot(index='', columns='', values='')
#### In Matrix Plot, we are not having Figure level function but only Axis Level function
#### Heatmap --> heatmap()
#### ---> data ---> matrix that has to be converted into an heatmap
#### ---> annot ---> it shows the numbers as well for the given rectangular grid
#### ---> cmap ---> A specific styling theme for the heatmap itself ---> coolwarms,virdis

```python
gap.head()
```

```python
dt=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(data=dt)
```

```python
dt=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(data=dt,annot=True,cmap='viridis',linewidth='0.5')
```
