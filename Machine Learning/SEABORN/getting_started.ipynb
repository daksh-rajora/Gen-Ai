## Relational Plot

#### Relational Plot ----> Bivariate Analysis
#### ---> Numerical - Numerical
#### ---> One Numerical Column ---> Change ---> How it Affects ---> Other Numerical Column
#### Line Plot ---> Line Plot is being used for Time Series Analysis (wrt Time how one quantity is going to vary) ---> X axis ---> Time Y axis --> Value that will vary
#### For ex --> wrt Time How my weight is varying or how sales value is varying with time or how stocl price for a company is varying with Time
#### Scatterplot --> Scatterplot show how 2 numerical variables vary or are correlated that means if one numerical column will vary then how it will affect the other numerical column
#### Correlation btw 2 Numerical Column

## Facet Plot
#### facet plot can help us divide the plot as per the multiple categories of a categorical column such as let us say, we dont want to use hue and we want to divide the plot or graph as per multiple categories of a categorical column as row wise or column wise then we can go with the facet plot
#### facet plot can help us separate the info Rowwise or Columnwise
#### We can use Row or column parameter to mention that if we want to separate the info rowwise or column wise
#### Facet plot functionality is only possible in the Figure level plot
#### As figure level plot provides us the control over the whole figure and axis level plot only provides us the control over the axis itself

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
sns.relplot(data=tips,x='total_bill',row='sex',y='tip',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',col='sex',y='tip',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',row='smoker',y='tip',kind='scatter',col='sex')
```

```python
gap.head()
```

```python
sns.relplot(data=gap,x='lifeExp',col='year',y='pop',kind='scatter',col_wrap=3)
```

### Distribution Plots

#### Distribution Plots provides us the idea about the distribution of data in a given column and typically in an Numerical Column to get to see that how the data is distributed in that column.
#### It will be used for the Univariate Analysis as we look into the distribution of the given Numerical Column.
#### We can find out the Outliers from the data as well
#### We can get the insight if we are having a normal distribution or Skewed and how the shape or Spread or Central Tendency of the Data is.
#### 1. Histplot --> Histplot is being used to check the distribution of a Numerical column wrt bins as the size of the bin reflects info for equal intervel

#### Axis Level Func ---> histplot()
#### Syntax--> histplot()
#### ---> data ---> dataframe where the column is
#### x---> column_name
#### bins---> number of bins i want inside my histogram(n)
#### kde ---> overlays a KDE curve
#### Figure Level Function
#### displot() ---> kind='hist'

```python
tips.head()
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
sns.histplot(data=tips,x='tip')
```

```python
gap.head()
```

```python
sns.histplot(data=gap,x='lifeExp')
```

```python
sns.histplot(data=tips,x='total_bill',bins=15,hue='sex',element='step')
```

```python
# Facet plot ---> figure level function ---> displot()
sns.displot(data=tips,x='total_bill',col='sex',kind='hist')
```

#### KDE(Kernel Density Estimation) Plot ---> To show the distribution of the data inside a Numerical Column itself but over here instead of getting bins, we can get the smooth curve which comes out as a probability density function representing the Distribution of the Data as with the curve How data is distributed exactly.

#### Axis Level Function ---> sns.kdeplot()
#### ---> data ---> dataframe
#### ---> x ---> column whose distribution we want to find
#### ---> fill ---> To fill under the curve ---> True or False ( By default --> False)

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex')
```

```python
sns.kdeplot(data=gap,x='lifeExp')
```

#### RugPlot ---> It adds the small vertical lines like ticks along the axis that show the individual data points.
#### More deep the rug plot tick ---> More density we can expect Their
#### It is usually used in conjuction with other plots like with kde plot as ---> kde Plot + RugPlot
#### Axis Level Function --->sns.rugplot()

```python
sns.rugplot(data=gap,x='lifeExp')
```

```python
sns.rugplot(data=tips,x='total_bill')
```
