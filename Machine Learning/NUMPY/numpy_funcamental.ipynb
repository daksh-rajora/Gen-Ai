## Numpy Arrays VS Python Sequences
#### Numpy Arrays have a fixed size at creation, unlike Python lists which grow dynamically.
#### Changing the size of ndim array will create a new array and delete the original. 
#### The elemets in a Numpy Arrau are required to be of the same type and thus will be of the same size in memory.
#### Numpy arrays facilitate advaned mathematical and other types of operations on large number of data. Typically, such operations are executed more efficiently with less code than is possible using Python's built in sequences.
