# NUMPY

#### Numpy is a Fundamental Package for scientific computing in Python. It is a Python Library that provides a Multi Dimentional Array Object and various derived objects such as Masked Arrays or Matrices and an assortment of routines for fast operations on Arrays including Mathematical, Logical, Shape Manipulation, Sorting, Selecting, I/O, Discreate Fourier Transform, Basic Linear Algebra, Stastical Operations, Random Simulation, etc.

#### At the core of the Numpy Package is the ndimensional Array Object.

Syntax---> Python----> Easy
## We needed Speed and Less Memory to work with the huge amount of data.

Numpy --> C language --> wrapped ---> Python Language

## Numpy---> Syntax---> Easy for beginners--> Performing Scientific Calculations---> Creating Multi Dimensional Arrays---> Speed is faster for applying Operations --> Less Memory

### Libraries like Scikit Learn, Tensor Flow and Pytorch require input data in Numpy Arrays or Tensors.

## Numpy Arrays VS Python Sequences
#### Numpy Arrays have a fixed size at creation, unlike Python lists which grow dynamically.
#### Changing the size of ndim array will create a new array and delete the original. 
#### The elemets in a Numpy Arrau are required to be of the same type and thus will be of the same size in memory.
#### Numpy arrays facilitate advaned mathematical and other types of operations on large number of data. Typically, such operations are executed more efficiently with less code than is possible using Python's built in sequences.
