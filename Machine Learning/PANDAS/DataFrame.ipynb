## Attributes

#### 1. shape ---> it provides the info about the (row,column) information about the dataset
#### syntax --<> df.shape

#### 2. dtypes --> It provides the idea about the datatype of all columns, so all columns with their name and their associated data type will be shown by using dtypes
#### syntax---> df.types

#### 3. index --> for a Dataframe with labelled index we can get the idea about the label just by using index attribute

#### 4. columns ---> it provides the idea about all of the column names of the give dataframe

#### 5. values --> it provides the idea about the values present row by row inside the dataframe. it always returns a 2D numpy array representing the values of column row by row

#### 6. ndim ---> it provides the idea about the dimension of the dataframe

#### 7. size ---> Total number of elements present inside the dataframe, that means all rows * all columns (Column 1 ---> 50 rows, column 2---> 50 rows .....)

# Functions

#### 1. head()--> to get to have the idea about the data that your dataset is storing as the top 5 rows, that means that starting from the top we can get the top 5 rows and along with that if i want that instead of top 5 rows i want to see a fixed no. of rows from the rows, that number i can pass inside the head() as a paramenter

#### 2. tail() ---> provide the insight about the dataframe from the botton so tail() can show us the bottom 5 rows and again based on the value you will be passing as the parameter, those many values you can get using the tail fn()

#### 3. sample() ---> provides the insight about the dataframe as the random values present inside the dataset and again based on how many values i want i can put the parameter as per that

#### 4. info() ---> info() provides the idea --->
#### 1. about the datatype of all columns
#### 2. Number of Missing values of all columns
#### 3. Dataframe storage in Memory

#### 5. describe() ---> it provides an idea as the mathematical summary of the data --> minimum, maximum, count, mean, std, 25%, 50%, 75%
