# Functions

#### 1. head()--> to get to have the idea about the data that your dataset is storing as the top 5 rows, that means that starting from the top we can get the top 5 rows and along with that if i want that instead of top 5 rows i want to see a fixed no. of rows from the rows, that number i can pass inside the head() as a paramenter

#### 2. tail() ---> provide the insight about the dataframe from the botton so tail() can show us the bottom 5 rows and again based on the value you will be passing as the parameter, those many values you can get using the tail fn()

#### 3. sample() ---> provides the insight about the dataframe as the random values present inside the dataset and again based on how many values i want i can put the parameter as per that

#### 4. info() ---> info() provides the idea --->
#### 1. about the datatype of all columns
#### 2. Number of Missing values of all columns
#### 3. Dataframe storage in Memory
