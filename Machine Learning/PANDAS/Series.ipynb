```python
import numpy as np
import pandas as pd
```

## Pandas Series is a cloumn of data in a Table like in Excel or Google Sheets with labels called as indexes attached to each value.

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'], name='Price of Grocery Items')
```

### Series is a 1D Array that can hold any type of data like integers, float, strings(object), boolean,mixed data type.

### Data type ---> Integers ---> Data in a Series(Column)--->Quantity of items in the Stock
### Data type ---> Float ---> Data in Series (Column) ---> Price or Product Ratings
### Data type ---> Strings ---> Column ---> Product Name
### Data type ---> Boolean ---> Column ---> Is item in stocl?
### Data type ---> Dates / Times ---> Column ---> Expiry Dates or sales Dates
### Mixed Types ---> Milk,40,True

## Creating a series in multiple ways
### 1. List
### 2. Numpy Array
### 3. Dictionary
### 4. Custom Index
### 5. Scalar Value

```python
# Series from list

country=['India','China','USA','Pakistan']

pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]
pd.Series(batsman_runs,dtype='int32')
```

```python
subject_names=['Math','Physics','Chemistry','Biology','English']
subject_marks=[34,54,23,45,64]

pd.Series(data=subject_marks,index=subject_names,name='Marks Scored' )
```

```python
## creating a series using numpy arrays
values=np.arange(10,21)
pd.Series(values)
```

```python
## Dictionaries to series

## Dictionaries is another data structure that stores the data in a key value pair and just because it 
# stores the data in akey value pair we dont have to explicitly mention the index

data_items={'Maths':78,'English':92,'Physics':76}
pd.Series(data_items)
```

```python
# custom index to series
values=[1,2,3,4,5,6]
pd.Series(data=values, index=['a','b','c','d','e','f'])
```

```python
# scalar value inside an series
pd.Series(4,
index=['a','b','c','d','e','f'])
```
