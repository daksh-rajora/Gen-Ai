## Pandas Series is a cloumn of data in a Table like in Excel or Google Sheets with labels called as indexes attached to each value.

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Butter'], name='Price of Grocery Items')
```

### Series is a 1D Array that can hold any type of data like integers, float, strings(object), boolean,mixed data type.

### Data type ---> Integers ---> Data in a Series(Column)--->Quantity of items in the Stock
### Data type ---> Float ---> Data in Series (Column) ---> Price or Product Ratings
### Data type ---> Strings ---> Column ---> Product Name
### Data type ---> Boolean ---> Column ---> Is item in stocl?
### Data type ---> Dates / Times ---> Column ---> Expiry Dates or sales Dates
### Mixed Types ---> Milk,40,True

```python
# Series from list

country=['India','China','USA','Pakistan']

pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]
pd.Series(batsman_runs,dtype='int32')
```

```python
subject_names=['Math','Physics','Chemistry','Biology','English']
subject_marks=[34,54,23,45,64]

pd.Series(data=subject_marks,index=subject_names,name='Marks Scored' )
```

```python
## creating a series using numpy arrays
values=np.arange(10,21)
pd.Series(values)
```

```python
## Dictionaries to series

## Dictionaries is another data structure that stores the data in a key value pair and just because it 
# stores the data in akey value pair we dont have to explicitly mention the index

data_items={'Maths':78,'English':92,'Physics':76}
pd.Series(data_items)
```

```python
# custom index to series
values=[1,2,3,4,5,6]
pd.Series(data=values, index=['a','b','c','d','e','f'])
```

```python
# scalar value inside an series
pd.Series(4,
index=['a','b','c','d','e','f'])
```

## Series Methods
#### 1. head()---> Provides the Top Preview of Data as the Top 5 Rows of the dataframe or series. It  is quite helpful with providing the preview of the dataset and if we want to preview top 10 rows then we can pass 10 as a argument or for top 7 rows we can pass 7 as an argument.

#### 2. tail()--> provides the bottom preview of Data as the Bottom 5 Rows.

#### for eg ---> dataset ---> salary column ---> top 5 rows ---> salary ---> 7 LPA ---> top 10 ---> top 15.... tail() ---> bottom 5 rows ---> salary ---> 4 LPA,9 LPA

#### 3. sample() ---> sample() provides any custom row from the dataset for us to preview it

#### 4. value_counts()---> it provides the idea about the count of the values (categorical values), if i am having a column where the categorical value count i want to calculate then i can go with value_counts()

#### 5. sort_values() --> this method sorts the value in an ascending manner. it can be useful for us fot the numerical columns. you can sort the value in descending manner as well

#### 6. sort_index() ---> this method sorts the index of the dataset.

## Mathematical Functions in Pandas
### 1. sum() --> sum of all values present in the series
### 2. product() ---> product of all values in the series
### 3. mean() ---> Mean provides us with the average value of the data column.
### for eg ---> Company data---> salary ---> average value for that column salary = 7,6,8,9,11 mean = (6+6+8+9+11)/5 ---> 8 LPA

### median()---> Median also provides the idea about the average of the data. for coming up with median, we need to SORT the values in Ascending manner and then the middle value after sorting the data --> median; data ---> salary---> 1 crore, 10 crore (Outlier) --> avg --> 7+8+6+9

### mode() --> mode provides us with the idea that which value is getting repeated maximum number of times.

### std()--> if i need to get the idea that how the values inside the series are distributed from the average or mean value then we can come up with the standard deviation
#### salary ---> 7,8,6,6.5,7.5 ---> low standard deviation
#### salary --> 7,15,4,20,28,17 ---> high standard deviation

### variance --> variance also provides the insight about the distribution of the data from the mean value and variance is the square of the standard deviation.

#### standard_deviation = |mean - v_1| + |mean - v_2| ..../total count of values
#### variance = std**2

### min() ---> minimum value in the series
### max() --> maximum value in the series
### describe() ---> describe function provides the idea about the summary of the important mathematical calculaations for the data.

#### min---> minimum value
#### max--> maximum value
#### mean--> average value
#### std --> standard deviation
#### 25%---> 255 of the values out of my all values are less than this specific value
#### 50% ---> 50% of the values out of all my values are less than this value
#### 75% ---> 75% of the values out of all my values are less than this specific value

#### count--> total number of rouws in the column

##### JEE mains ---> 25% ---> my score in jee maiins was greater than 25% of the total applicants who applied in jee
